# Python `time` and `datetime` Modules: Study Guide

Python has powerful built-in modules to handle dates, times, and calendars. The most commonly used module is `datetime`, which is very feature-rich.

## 1. Core Classes in the `datetime` Module

The `datetime` module is divided into several specific classes (referred to as categories/chapters in the transcript):

* **`datetime`**: Combines both date (year, month, day) and time (hours, minutes, seconds, microseconds).
* **`date`**: Contains only date information (year, month, day).
* **`time`**: Contains only time information (hours, minutes, seconds, microseconds). *Note: A day has 24 hours, an hour has 60 minutes, and a minute has 60 seconds.*
* **`timedelta`**: Represents the difference or duration between two dates or times (measured in days, seconds, and microseconds).
* **`tzinfo`**: Provides time zone information (e.g., Indian Standard Time, GMT, etc.) and handles Daylight Saving Time (DST).

---

## 2. The `time` Module and "The Epoch"

Any computer clock needs a starting point. In Python (and most operating systems), this starting point is called the **Epoch**, which is **January 1, 1970, at 12:00 AM**.

### `time.time()`

This function returns the total number of seconds that have passed since the Epoch.

```python
import time

seconds_passed = time.time()
print(seconds_passed) # E.g., 1647340000.0 seconds

```

### `time.localtime()`

This function takes the seconds since the Epoch and converts them into a `struct_time` object, breaking it down into year, month, day, hours, minutes, seconds, etc.

```python
import time

lt = time.localtime()
print(lt.tm_year)  # Returns the current year (e.g., 2022)
print(lt.tm_mon)   # Returns the current month (1-12)
print(lt.tm_mday)  # Returns the day of the month
print(lt.tm_wday)  # Returns the day of the week (0 = Monday, 1 = Tuesday...)

```

*Note on Daylight Saving Time (DST):* `localtime()` also checks if Daylight Saving Time is active. DST is a practice where clocks are moved forward in summer (giving more daylight in the evening) and backward in winter. Countries near the equator (like India) do not observe DST.

**The 1996 Example:**
If you take the current `time.time()` (which in 2022 was 52 years of seconds since 1970) and divide it by 2, you get 26 years. If you pass those seconds into `time.localtime()`, it will return a date in **1996** (1970 + 26 years).

### `time.ctime()`

This function returns the current date and time as a highly readable **string**.

```python
import time
print(time.ctime()) # Output: 'Tue Mar 15 16:41:03 2022'

```

## 5. Mathematical Operations and Comparisons

You can use mathematical operators to compare dates or find the difference between them.

### Comparing Objects (>, <, ==)

You can compare `date`, `time`, and `datetime` objects to see which one is older or newer.

```python
dt1 = datetime.datetime(2010, 1, 1)
dt2 = datetime.datetime(2011, 1, 1)

print(dt2 > dt1) # Returns True (2011 is newer/greater than 2010)

```

### Subtracting Objects

You can subtract two `date` or `datetime` objects to get the exact difference between them. The result is a `timedelta` object.

```python
d1 = datetime.date(2015, 1, 1)
d2 = datetime.date(2016, 1, 1)

difference = d2 - d1
print(difference.days) # Returns 365

```

**Crucial Rule for Subtraction:**

* `date - date` = Valid (`timedelta`)
* `datetime - datetime` = Valid (`timedelta`)
* `time - time` = **NOT Supported.** You cannot subtract two isolated `time` objects because Python doesn't know what days they belong to (e.g., are they 2 hours apart, or 24 hours and 2 hours apart?).

# Python Date and Time Formatting: `strftime` and `strptime`

When working with dates and times in Python, you rarely want to display them in their raw object format (e.g., `2010-01-01 10:20:30`). Instead, you usually want to format them nicely for users (e.g., `"01 January, 2010"`) or take a string provided by a user and turn it into a Python object.

Python provides two essential methods for this:

1. **`strftime` (String Format Time):** Converts a Python `date`, `time`, or `datetime` object into a formatted **string**.
2. **`strptime` (String Parse Time):** Converts a formatted **string** into a Python `datetime` object.

---

### Key Format Codes (Control Characters)

Both methods use special "format codes" (prefixed with a `%`) to represent different parts of a date and time. Here are the ones mentioned in your lesson:

| Code | Description | Example |
| --- | --- | --- |
| **`%d`** | Day of the month (zero-padded) | `01`, `03`, `15` |
| **`%-d`** | Day of the month (not zero-padded)* | `1`, `3`, `15` |
| **`%b`** | Month abbreviation | `Jan`, `Feb` |
| **`%B`** | Full month name | `January`, `February` |
| **`%y`** | Year without century (2 digits) | `10`, `22` |
| **`%Y`** | Year with century (4 digits) | `2010`, `2022` |
| **`%a`** | Weekday abbreviation | `Mon`, `Fri` |
| **`%A`** | Full weekday name | `Monday`, `Friday` |
| **`%H`** | Hour (24-hour clock) | `00` to `23` |
| **`%I`** | Hour (12-hour clock) | `01` to `12` |
| **`%M`** | Minute | `00` to `59` |
| **`%S`** | Second | `00` to `59` |

*(Note: The `-` modifier, like `%-d` or `%-H`, removes leading zeros. It works on Linux/Mac. If you are on Windows, the equivalent is `#`, like `%#d`).*

---

### 1. `strftime`: Object to String

The `strftime()` method is available on `date`, `time`, and `datetime` objects. You pass it a string containing the format codes you want to use. Any normal characters (like commas, spaces, or colons) will be printed exactly as you write them.

**Code Example:**

```python
import datetime

# 1. Create a datetime object (Year 2010, Jan 1st, 10:20:30)
dt1 = datetime.datetime(2010, 1, 1, 10, 20, 30)

# 2. Format using uppercase letters (Full names, zero-padded, 4-digit year)
format_1 = dt1.strftime("%d %B, %A %Y %H:%M:%S")
print(format_1)  
# Output: 01 January, Friday 2010 10:20:30

# 3. Format using lowercase/unpadded letters (Abbreviations, 2-digit year)
format_2 = dt1.strftime("%-d %b, %a %y %H:%M:%S")
print(format_2)  
# Output: 1 Jan, Fri 10 10:20:30

```

---

### 2. `strptime`: String to Object

The `strptime()` method is a class method belonging to the `datetime.datetime` class. It takes two arguments:

1. The **string** containing the date/time.
2. The **format code string** that tells Python exactly how to read that string.

**Crucial Rule:** The format string you provide *must perfectly match* the structure of the text string, including all spaces, commas, and colons.

**Code Example:**

```python
import datetime

# 1. We have a date/time in the form of a string
str_date = "10 Jan 2015, 10:15:30"

# 2. We map the format codes to match the string exactly
# "10" = %d
# "Jan" = %b
# "2015" = %Y
# ", " = literal comma and space
# "10:15:30" = %H:%M:%S
dt2 = datetime.datetime.strptime(str_date, "%d %b %Y, %H:%M:%S")

# 3. Now it is a real Python datetime object!
print(dt2)        # Output: 2015-01-10 10:15:30
print(dt2.year)   # Output: 2015
print(dt2.minute) # Output: 15

```

# 📘 Python Date and Time: The `timedelta` Class

When working with dates and times in Python, you don't just want to record specific moments in time; you frequently need to calculate the **duration** between two moments or calculate future/past dates. 

The `timedelta` class (part of the built-in `datetime` module) is specifically designed to represent a period, duration, or difference in time.

## 1. What is `timedelta`?
A `timedelta` object represents a duration of time. Internally, Python strictly stores this duration using only three units:
*   **Days**
*   **Seconds**
*   **Microseconds**

Even if you define a `timedelta` using hours or weeks, Python converts it down to these three internal units. Noticeably missing are "Months" and "Years"—this is because the length of a month or year varies, making them mathematically inconsistent for strict duration calculations.

## 2. Generating a `timedelta` via Subtraction
When you subtract one `datetime` (or `date`) object from another, Python automatically returns a `timedelta` object representing the exact time that elapsed between them.

```python
from datetime import datetime

# Define an older date and a newer date
dt1 = datetime(2010, 10, 1, 10, 20, 30)
dt2 = datetime(2011, 10, 1, 10, 20, 30)

# Subtracting dates yields a timedelta object
duration = dt2 - dt1

print(type(duration)) # Output: <class 'datetime.timedelta'>
print(duration.days)  # Output: 365

```

## 3. Extracting Human-Readable Durations (`divmod`)

Because `timedelta` only gives you total days or total seconds, you often need to calculate human-readable chunks like "Weeks" or "Months" (assuming a 30-day average month) manually.

Python's built-in `divmod(a, b)` function is perfect for this. It returns a tuple containing the **quotient** (division result) and the **remainder** (modulo result).

```python
from datetime import datetime

duration_days = 365

# 1. Calculate rough months (assuming 30 days/month)
months, leftover_days = divmod(duration_days, 30)
print(f"{months} months and {leftover_days} days") 
# Output: 12 months and 5 days

# 2. Calculate exact weeks
weeks, leftover_days = divmod(duration_days, 7)
print(f"{weeks} weeks and {leftover_days} days") 
# Output: 52 weeks and 1 days

```

## 4. Date Arithmetic: Creating Future or Past Dates

You can manually create a `timedelta` object and add it to (or subtract it from) a `datetime` object to calculate a date in the future or the past.

When creating a `timedelta`, you can pass arguments for `days`, `seconds`, `microseconds`, `milliseconds`, `minutes`, `hours`, and `weeks`.

```python
from datetime import datetime, timedelta

# Get the current date and time
now = datetime.now()
print("Current Time:", now) 
# Example Output: 2026-07-26 17:21:19.000000

# 1. Create a duration of 31 days
duration_31_days = timedelta(days=31)

# 2. Add the duration to the current time to get a future date
future_date = now + duration_31_days
print("31 Days Later:", future_date)
# Example Output: 2026-08-26 17:21:19.000000

# 3. Go backwards in time
past_date = now - timedelta(weeks=2)
print("2 Weeks Ago:", past_date)

```


# Python `calendar` Module: Study Guide

The built-in `calendar` module allows you to output calendars like the Unix `cal` program and provides additional useful functions related to the calendar. By default, these calendars follow the European convention where **Monday is the first day of the week (0) and Sunday is the last (6).**

## 1. Basic Calendar Functions

You can use the `calendar` module directly without creating any objects for simple tasks.

```python
import calendar

# 1. First Day of the Week
print(calendar.firstweekday()) 
# Output: 0 (Which represents Monday)

# You can change this to Sunday (6) if you prefer:
calendar.setfirstweekday(6)

# 2. Check for Leap Years
print(calendar.isleap(2024)) 
# Output: True

# 3. Count Leap Years within a Range (start is inclusive, end is exclusive)
print(calendar.leapdays(2000, 2022)) 
# Output: 6 

# 4. Find the Day of the Week for a Specific Date (Year, Month, Day)
# 0 = Monday, 1 = Tuesday ... 5 = Saturday, 6 = Sunday
print(calendar.weekday(2000, 1, 1)) 
# Output: 5 (Saturday)

```

## 2. Printing Simple Calendars

You can quickly print a formatted calendar for a specific month or year directly to the console.

```python
import calendar

# Print a specific month
print(calendar.month(2022, 3))

# You can adjust column width (w) and line spacing (l)
print(calendar.month(2022, 4, w=5, l=2)) 

```

---

## 3. The `Calendar` Class (Iterators)

If you want to extract date data programmatically (rather than just printing it to the screen), you can create a `Calendar` object. This provides **iterators** (referred to in the transcript as "repeaters") to loop through dates.

```python
import calendar

# Create a calendar object (0 means weeks start on Monday)
cal = calendar.Calendar(0)

# 1. Loop through the days of the week (0 to 6)
for day in cal.iterweekdays():
    print(day, end=" ")
# Output: 0 1 2 3 4 5 6 

print("\n---")

# 2. Iterate through all dates in a month
# NOTE: This will include a few days from the previous/next month 
# to complete the full 7-day week grid!
for dt in cal.itermonthdates(2022, 2):
    print(dt)
# Output: 2022-01-31 (Monday)
#         2022-02-01 (Tuesday) ... up to the end of the week in March.

# 3. Get the month as a list of weeks (each week is a list of datetime.date objects)
weeks = cal.monthdatescalendar(2022, 3)
print(weeks[0]) # Prints the first week of March

```

### 🧠 Summary / Interview Prep

* **What is the difference between `calendar.month()` and `calendar.prmonth()`?**
* `calendar.month()` returns a formatted *string* that you must pass into a `print()` function.
* `calendar.prmonth()` is a method on a `TextCalendar` object that directly prints the calendar to the console without needing a `print()` statement.


* **Why does `itermonthdates(2022, 2)` (February) return a date from January?**
* Calendar grids are designed in complete 7-day rows. If February 1st falls on a Tuesday, the calendar must fill the Monday slot with January 31st to complete the visual grid of that week. Python's iterator reflects this real-world calendar behavior.
